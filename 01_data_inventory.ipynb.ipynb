```python
# Load all the 9 Datasets into Python

import pandas as pd
from pathlib import Path

# Project folder
project_folder = Path(
    r"C:\Users\vinaya kumar\OneDrive\문서\Indian_E-commerce_Business_Intelligence"
)

# Raw dataset folder
raw_folder = project_folder / "-Raw_Dataset"

# Load all datasets
customers = pd.read_csv(raw_folder / "customers.csv")
orders = pd.read_csv(raw_folder / "orders.csv")
order_items = pd.read_csv(raw_folder / "order_items.csv")
products = pd.read_csv(raw_folder / "products.csv")
payments = pd.read_csv(raw_folder / "payments.csv")
shipments = pd.read_csv(raw_folder / "shipments.csv")
returns = pd.read_csv(raw_folder / "returns.csv")
customer_reviews = pd.read_csv(raw_folder / "customer_reviews.csv")
marketing_campaigns = pd.read_csv(raw_folder / "marketing_campaigns.csv")

print("All 9 datasets loaded successfully!")
```

```python
# how many rows and columns each dataset contains..

datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Products": products,
    "Payments": payments,
    "Shipments": shipments,
    "Returns": returns,
    "Customer Reviews": customer_reviews,
    "Marketing Campaigns": marketing_campaigns
}

inventory = []

for name, df in datasets.items():
    inventory.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    })

inventory_df = pd.DataFrame(inventory)

inventory_df
```

```python
# Save the inventory report.

inventory_df.to_csv(
    project_folder / "Project_Reports" / "data_inventory.csv",
    index=False
)

print("Data inventory report saved successfully!")
```

```python
#  Each column contains text, numbers, dates or Boolean values.

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print("DATASET:", name)
    print("=" * 60)

    print(df.dtypes)
```

```python
# Check Missing Values.


missing_report = []

for name, df in datasets.items():

    for column in df.columns:

        missing_count = df[column].isna().sum()

        if missing_count > 0:

            missing_report.append({
                "Dataset": name,
                "Column": column,
                "Missing Values": missing_count,
                "Missing Percentage": round(
                    (missing_count / len(df)) * 100, 2
                )
            })

missing_df = pd.DataFrame(missing_report)

missing_df
```

```python
 # Investigate Missing values

# Investigation 1: Customers with missing last order date

missing_last_order = customers[
    customers["last_order_date"].isna()
]

print("Customers with missing last order date:",
      len(missing_last_order))

print("\nCustomer status:")
print(missing_last_order["customer_status"].value_counts())

print("\nTotal orders:")
print(missing_last_order["total_orders"].value_counts())


# Investigation 2: Shipments with missing delivery date

missing_delivery = shipments[
    shipments["actual_delivery_date"].isna()
]

print("\nShipments with missing actual delivery date:",
      len(missing_delivery))

print("\nDelivery status:")
print(missing_delivery["delivery_status"].value_counts())
```

```python
# Check the duplicate rows in all nine datasets.

duplicate_report = []

for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    duplicate_report.append({
        "Dataset": name,
        "Total Records": len(df),
        "Duplicate Rows": duplicate_count
    })

duplicate_df = pd.DataFrame(duplicate_report)

display(duplicate_df)
```

```python
# Save the report

duplicate_df.to_csv(
    project_folder / "Project_Reports" / "duplicate_report.csv",
    index=False
)

print("Duplicate report saved successfully!")
```

```python
# check the each table's primary key is unique.

primary_keys = {
    "Customers": ("customers", "customer_id"),
    "Orders": ("orders", "order_id"),
    "Order Items": ("order_items", "order_item_id"),
    "Products": ("products", "product_id"),
    "Payments": ("payments", "payment_id"),
    "Shipments": ("shipments", "shipment_id"),
    "Returns": ("returns", "return_id"),
    "Customer Reviews": ("customer_reviews", "review_id"),
    "Marketing Campaigns": ("marketing_campaigns", "campaign_id")
}

primary_key_report = []

for table_name, (variable_name, column) in primary_keys.items():

    df = datasets[table_name]

    primary_key_report.append({
        "Dataset": table_name,
        "Primary Key": column,
        "Total Records": len(df),
        "Unique Values": df[column].nunique(),
        "Duplicate Keys": df[column].duplicated().sum(),
        "Missing Keys": df[column].isna().sum()
    })

primary_key_df = pd.DataFrame(primary_key_report)

display(primary_key_df)
```

```python
foreign_keys = [
    ("Orders", "customer_id", "Customers", "customer_id"),
    ("Order Items", "order_id", "Orders", "order_id"),
    ("Order Items", "product_id", "Products", "product_id"),
    ("Payments", "order_id", "Orders", "order_id"),
    ("Shipments", "order_id", "Orders", "order_id"),
    ("Returns", "order_id", "Orders", "order_id"),
    ("Returns", "customer_id", "Customers", "customer_id"),
    ("Returns", "product_id", "Products", "product_id"),
    ("Customer Reviews", "order_id", "Orders", "order_id"),
    ("Customer Reviews", "customer_id", "Customers", "customer_id"),
    ("Customer Reviews", "product_id", "Products", "product_id")
]

foreign_key_report = []

for child, child_col, parent, parent_col in foreign_keys:

    child_df = datasets[child]
    parent_df = datasets[parent]

    invalid_count = (
        ~child_df[child_col].isin(parent_df[parent_col])
    ).sum()

    foreign_key_report.append({
        "Child Dataset": child,
        "Foreign Key": child_col,
        "Parent Dataset": parent,
        "Parent Key": parent_col,
        "Invalid References": invalid_count
    })

foreign_key_df = pd.DataFrame(foreign_key_report)

display(foreign_key_df)
```

```python
# to save the Validation Summary.

validation_summary = pd.DataFrame([
    {
        "Validation": "Duplicate Rows",
        "Total Checks": len(duplicate_df),
        "Failed Checks": (duplicate_df["Duplicate Rows"] > 0).sum()
    },
    {
        "Validation": "Primary Keys",
        "Total Checks": len(primary_key_df),
        "Failed Checks": (
            (primary_key_df["Duplicate Keys"] > 0) |
            (primary_key_df["Missing Keys"] > 0)
        ).sum()
    },
    {
        "Validation": "Foreign Keys",
        "Total Checks": len(foreign_key_df),
        "Failed Checks": (
            foreign_key_df["Invalid References"] > 0
        ).sum()
    }
])

display(validation_summary)

validation_summary.to_csv(
    project_folder / "Project_Reports" / "validation_summary.csv",
    index=False
)
```

```python
# Validate the business rule.

business_checks = {
    "Customer age is positive": (customers["age"].dropna() > 0).all(),

    "Order amounts are non-negative": (
        (orders[["subtotal", "shipping_fee", "tax_amount", "final_amount"]] >= 0)
        .all().all()
    ),

    "Order item quantities are positive": (
        (order_items["quantity"] > 0).all()
    ),

    "Product prices are positive": (
        (products[["price", "cost_price"]] > 0).all().all()
    ),

    "Product cost does not exceed price": (
        (products["cost_price"] <= products["price"]).all()
    ),

    "Ratings are between 1 and 5": (
        products["rating_average"].dropna().between(1, 5).all()
        and customer_reviews["rating"].between(1, 5).all()
    ),

    "Order discounts are between 0 and 100": (
        orders["discount_percentage"].between(0, 100).all()
    ),

    "Item discounts are between 0 and 100": (
        order_items["discount_percentage"].between(0, 100).all()
    ),

    "Shipment delay flags are 0 or 1": (
        shipments["delayed_flag"].isin([0, 1]).all()
    )
}

business_rule_report = pd.DataFrame([
    {
        "Business Rule": rule,
        "Status": "PASS" if passed else "FAIL"
    }
    for rule, passed in business_checks.items()
])

display(business_rule_report)
```

```python
# Save the overall the validation results.

# let's combine our four validation reports into one final data quality summary.

overall_validation = pd.DataFrame([
    {
        "Category": "Duplicate Rows",
        "Total Checks": 9,
        "Passed": 9,
        "Failed": 0
    },
    {
        "Category": "Primary Keys",
        "Total Checks": 9,
        "Passed": 9,
        "Failed": 0
    },
    {
        "Category": "Foreign Keys",
        "Total Checks": 11,
        "Passed": 11,
        "Failed": 0
    },
    {
        "Category": "Business Rules",
        "Total Checks": len(business_rule_report),
        "Passed": (business_rule_report["Status"] == "PASS").sum(),
        "Failed": (business_rule_report["Status"] == "FAIL").sum()
    }
])

display(overall_validation)

overall_validation.to_csv(
    project_folder / "Project_Reports" / "overall_validation_report.csv",
    index=False
)

print("Overall validation report saved successfully!")
```

```python
# DATA CLEANING.

# Create the cleaned datasets folder path.

cleaned_folder = project_folder / "Cleaned_ Dataset"

print("Cleaned folder exists:", cleaned_folder.exists())
```

```python
print("Ready to create cleaned datasets!")
```

```python
# Convert date columns into Datetime formats.

# Define date columns for each dataset

date_columns = {
    "Customers": ["customer_signup_date", "last_order_date"],
    "Orders": ["order_date"],
    "Products": ["product_launch_date"],
    "Payments": ["payment_date"],
    "Shipments": [
        "dispatch_date",
        "expected_delivery_date",
        "actual_delivery_date"
    ],
    "Returns": ["return_date"],
    "Customer Reviews": ["review_date"],
    "Marketing Campaigns": [
        "campaign_start_date",
        "campaign_end_date"
    ]
}

# Convert date columns

for dataset_name, columns in date_columns.items():

    df = datasets[dataset_name]

    for column in columns:
        df[column] = pd.to_datetime(
            df[column],
            errors="coerce"
        )

print("Date conversion completed!")
```

```python
# errors= "coerce" helps to converts invalid date values into NaT(Missing Date time Values).
```

```python
for dataset_name, columns in date_columns.items():

    print("\n", dataset_name)

    for column in columns:
        print(column, ":", datasets[dataset_name][column].dtype)
```

```python
# Verify missing values after conversion.

date_missing_report = []

for dataset_name, columns in date_columns.items():

    df = datasets[dataset_name]

    for column in columns:

        missing_count = df[column].isna().sum()

        date_missing_report.append({
            "Dataset": dataset_name,
            "Date Column": column,
            "Missing Values": missing_count
        })

date_missing_df = pd.DataFrame(date_missing_report)

display(date_missing_df)
```

```python
# Save the Date missing values report.

date_missing_df.to_csv(
    project_folder / "Project_Reports" / "date_missing_report.csv",
    index=False
)

print("Date missing-value report saved!")
```

```python
# Handle remaining missing values.

remaining_missing = []

for name, df in datasets.items():

    for column in df.columns:

        missing_count = df[column].isna().sum()

        if missing_count > 0:
            remaining_missing.append({
                "Dataset": name,
                "Column": column,
                "Missing Values": missing_count
            })

remaining_missing_df = pd.DataFrame(remaining_missing)

display(remaining_missing_df)
```

```python
# customers(age) Datasets   -  keep as missing.
# customers(preferred_device) -  replace with unknown.
# customers(last_order_Date)  -  keep as missing.
# orders(coupon code)   -     replace with NO coupon.
# orders(campaign_id)   -     replace with no campaign.
# products(rating_Average)-   keeping as missing.
# shipments(actual_delievery_date) - keep as missing.
# shhipments(delievery_Days) - keep as missing.
# customer reviews( review_sentiments)  -  replace with unknown.
```

```python
# Customers
customers["preferred_device"] = (
    customers["preferred_device"].fillna("Unknown")
)

# Orders
orders["coupon_code"] = (
    orders["coupon_code"].fillna("No Coupon")
)

orders["campaign_id"] = (
    orders["campaign_id"].fillna("No Campaign")
)

# Customer Reviews
customer_reviews["review_sentiment"] = (
    customer_reviews["review_sentiment"].fillna("Unknown")
)

print("Missing-value treatment completed!")
```

```python
# verify the results.

print("Customers - preferred_device:")
print(customers["preferred_device"].isna().sum())

print("\nOrders - coupon_code:")
print(orders["coupon_code"].isna().sum())

print("\nOrders - campaign_id:")
print(orders["campaign_id"].isna().sum())

print("\nCustomer Reviews - review_sentiment:")
print(customer_reviews["review_sentiment"].isna().sum())
```

```python
# Final Missing values.

final_missing_report = []

for name, df in datasets.items():

    missing_count = df.isna().sum().sum()

    final_missing_report.append({
        "Dataset": name,
        "Remaining Missing Values": missing_count
    })

final_missing_df = pd.DataFrame(final_missing_report)

display(final_missing_df)
```

```python
# Export the cleaned dataframes into the Cleaned datasets.

# Export all nine cleaned datasets

for name, df in datasets.items():

    file_name = name.lower().replace(" ", "_") + ".csv"

    df.to_csv(
        cleaned_folder / file_name,
        index=False
    )

    print(f"Exported: {file_name}")

print("\nAll 9 cleaned datasets exported successfully!")
```
